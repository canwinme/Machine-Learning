```python
f = pd.Series({'사과': 1200, '딸기':3500, '키위':2100, '메론':9600})
print(f)
print(f[2])
print(f['키위'])
print(f[1:4])
print(f['딸기':'메론'])
```

```python
f = pd.Series([1200, None, 2100, 9600], index=['사과','딸기','키위','메론'])
print(f)
print()
print(f.isnull()) ## 전체 출력
print()
print(f[f.isnull()]) ##인덱스만 출력
print()
print(f.drop('딸기')) ## 복사본만 삭제 원본은 그대로
print()
print(f.drop('딸기', inplace = True)) ##inplace = True 를 사용하면 원본도 삭제됨.
print(f)
```

* 빈셀을 처리하는 일반적인 방법은 평균, 중앙값, 최빈값을 계산한다.
* mean(): 평균. median(): 중앙값, mode(): 최빈값


```python
file_path = '/content/drive/MyDrive/test2.csv'
df = pd.read_csv(file_path, encoding= 'cp949')
print(df)
print()
avg = df['몸무게'].mean()
print(avg)
df['몸무게']= df['몸무게'].fillna(avg)
print(df)
med = df['나이'].median()
print(med)
df['나이']= df['나이'].fillna(med)
print(df)
```

```python
print('최대값=', df['급여'].max())
print('최소값=', df['급여'].min())
print('합계=', df['급여'].sum())
print('평균=', df['급여'].mean())
print('중앙값=', df['급여'].median())
print('최빈값=', df['급여'].mode())
print()
print('데이터 갯수=', df['종목'].count())
print('값 개수=', df['종목'].value_counts())
print('통계량=', df['급여'].describe()) #통계량 관련
print('다중집계=', df['급여'].agg('mean'))
```
