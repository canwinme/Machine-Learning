##선그래프

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 5)
y = np.arange(10, 50, 10)

plt.plot(x, y)
plt.show()
```

```python
x = np.array([1, 2, 3 ,4])
y = np.array([2, 4, 12, 14])

plt.plot(x, y, '^') #마커 모양 ^, o
plt.show()
```

# 다중선 그래프


```python
x = np.array([1,2,3,4])
y1 = np.array([2, 4, 12 ,14])
y2 = np.array([3, 5, 13 , 15])

plt.plot(x,y1, label ='first')
plt.plot(x,y2, label = 'second')
plt.legend() # 범례를 표시할 때 사용
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.rc('font', family ='NanumGothic')
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="whitegrid", font="NanumGothic")

plt.title('그래프 제목') #한글이 잘안됨 그래서 별도로 설치를해야함.
plt.plot([-20, -10, 10, 20])
plt.show()
```

```python
plt.title('그래프 제목', loc='right', fontsize=14, fontweight ='bold', color='blue', pad=50)
plt.plot([-20, -10, 10 , 20])
plt.show()
```

.grid() 메서드의 axis = 매개변수를 사용하여 표시할 그리드선을 지정할 수 있다.
* axis ='x
* axis ='y'
* axis = 'both'


```python
plt.plot([1,2,3,4], [28,25,26,21])
plt.grid(True)
plt.show()
```

* color='k' 그리드의 색상('b','g','r','c','m','y','k','w')

```python
from matplotlib.lines import lineStyles
plt.plot([1,2,3,4],[28,25,26,21])
linestyle='--'
linewidth =0.5 선 너비
alpha = 0.3 선 투명도

plt.grid(color='g', linestyle='--', linewidth=0.5, alpha =0.3)
plt.show()
```

```python
plt.plot([1,2,3,4], [28,25,26,21])
plt.grid(which='major', color = 'black', linestyle ='-', linewidth = 0.75) #which major 기본선

plt.minorticks_on()
plt.grid(which ='minor', color ='gray', linestyle=':', linewidth=0.5, alpha=0.5) #which minor 보조그리드선
```

```python
x = [1,2,3,4]
y = [77,78,79,80]

plt.plot(x,y)
plt.xlabel('주간' ,loc='right') #loc로 위치변경가능
plt.ylabel('체중', loc ='top') #bottom
plt.show()
```

```python
x = [1, 2, 3, 4]
y = [77, 78, 79, 80]

plt.plot(x, y)
plt.xlim(left = 0, right= 5)
plt.ylim(bottom = 10, top = 100)
plt.show()
```
