비선형 군집화: DBSCAN과 계층적 군집화
- DBSCAN:
  - eps: 내주변을 탐색하는 이웃 반경의 크기
  - min_samples: 내 반병안에 최소 몇명의 데이터가 모여있어야 하는가?
  - 3가지 점
    - 핵심점 : 반경내 min_samples 이상의 이웃을 품고있는 대장 점
    - 경계점 : 대장 점의 반경 안에는 들어와 있지만 내 주변에는 기준 인원이 부족한 외곽점
    - 노이즈점 : 이웃도 없는 아웃사이드 점(DBSCAN은 이를 자동으로 -1번 사물함에 버린다

- 계층적 군집화
  - 처음에는 모든데이터가 각각 독립된 1인용 방을 쓰다가 가장 가까운 것들끼리 손을 잡고 점차 큰방으로 합쳐져 나가는 방식
  - 링키지(연결기준) 종류:
    - ward : 군집 간의 분산증가를 최소화하는 방향으로 합침(가장 많이 쓰임)
    - single/complete : 가장 가까운 점끼리 또는 가장 먼 점끼리 거리 측정
    

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans
# 1. 반달 모양의 가상 데이터셋 만들기
X, _ = make_moons(n_samples=400, noise=0.05,random_state=42)
# 2. K-Means 모델로 2개 그룹 군집화
kmeans = KMeans(n_clusters=2, random_state=42, n_init='auto')
cluster_labels=kmeans.fit_predict(X)

#. 군집화 결과 시각화
plt.figure(figsize=(10,6))
# 데이터 포인터 시각화(K-Means 가 분류한 라벨대로 색상 결정.)
sns.scatterplot(x=X[:, 0],y=X[:, 1], hue=cluster_labels, palette ='Set2', s= 50)
# K-Means 가 지정한 중심점 표시
centriods = kmeans.cluster_centers_
plt.scatter(centriods[:,0],centriods[:,1],c='red',marker='X',s =200, label='Centroids')
plt.title('K-Means Fails on Moon_shapes Data')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN, AgglomerativeClustering
# 1. 초승달 비선형 데이터 생성
X, _ = make_moons(
    n_samples=400,
    noise=0.05,
    random_state=42
)
# 2. DBSCAN 모델 학습
dbscan = DBSCAN(eps=0.2, min_samples=7) #반경 0.2, 최소 7개
dbscan_labels = dbscan.fit_predict(X)

print(np.unique(dbscan_labels))
# 3. 계층 군집화 모델 학습
hierarchical = AgglomerativeClustering(n_clusters=2)   # 분산 최소화 방
hier_labels = hierarchical.fit_predict(X)
# 4. 결과 시각화 비교
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
# 첫번째 DBSCAN 결과
axes[0].scatter(X[:, 0], X[:, 1], c=dbscan_labels, cmap='viridis', s=60, edgecolor='k')
axes[0].set_title('DBSCAN (Density-Based)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('X1', fontsize=11)
axes[0].set_ylabel('X2', fontsize=11)
# 두번째 계층적 군집화 결과
axes[1].scatter(X[:, 0], X[:, 1], c=hier_labels, cmap='viridis', edgecolor='k')
axes[1].set_title('Hierarchical Clustering (Ward)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Feature1', fontsize=11)
axes[1].set_ylabel('Feature2', fontsize=11)
plt.tight_layout()
plt.show()
```
