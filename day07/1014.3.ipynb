최적의 군집찾기 : 엘보우 기법과 실루엣 분석

K를 어떻게 찾느냐?
- 엘보우 기법: 오차가 급격히 꺾이는 지점 찾기
  - 이너셔(inertia) : 각 데이터에서 자기가 속한 중심점까지의 거리 제곱의 합. 군집내부의 응집도를 나타내며, 이너셔가 작을수록 데이터들이 중심점에 바짝모여 있다는 것을 나타낸다.
  - K를 늘리면 이너셔가 작아진다. K를 1부터 늘려갈때 이너셔가 확줄어들다가 갑자기 완만해지는 구간을 만난다. 이지점이 최적의 K가 된다.
- 실루엣 분석: 내가 속한 무리와는 가깝고, 다른무리와는 얼마나 멀리 떨어져 있나?

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X, y_true = make_blobs(
    n_samples=300,
    #centers=3,
    cluster_std=0.6,
    random_state=42
)
inertia_list =[]
k_range = range(1,10)

for k in k_range:
  #각 K마다 K-Means 모델 생성 및 학습
  kmeans = KMeans(n_clusters = k, random_state=42, n_init=10)
  kmeans.fit(X)
  #모델의 inertia_ 속성값을 리스트에 저장한다.
  inertia_list.append(kmeans.inertia_)

plt.figure(figsize=(9,4))
plt.plot(k_range, inertia_list, marker ='o', color='royalblue',linewidth=2,markersize=8)
plt.title('The Elbow Method(Optimal K Detection)',fontsize =13,fontweight='bold' )
plt.xlabel('Number of clusters',fontsize =11)
plt.ylabel('Inertia(Within-Cluster sum of Squares)',fontsize=11)
plt.show()
```

```python
#실루엣 점수를 이용한 객관적 검증
print( "k에 따른 실루엣 점수 결과")
for k in range(2,6):
  kmeans = KMeans(n_clusters=k, random_state= 42, n_init=10)
  cluster_labels = kmeans.fit_predict(X)

  score = silhouette_score(X, cluster_labels)
  print(f"-> k= {k}일 때의 평균 실루엣 점수: {score:.4f}")
```

출력 결과에서 K=3일 때 실루엣점수가 가장 높거나 우수하게 나타남을 알 수 잇다.

이를 통해 K=3이 데이터의 군집을 가장 효과적으로 구분하는 최적의 군집 개수임을 확인할 수 있다. 따라서 K-Means 모델의 군집 개수를 3으로 설정하는 것이 적절하다고 판단된다.  
알 수 있는 점: K=3일 때 실루엣 점수가 가장 높으므로, 군집 내부의 데이터는 서로 유사하고 다른 군집과는 잘 구분된다는 것을 알 수 있다.
