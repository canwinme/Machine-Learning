베르누이 분포(이산확률분포)  ->2개의 결과를 같는 단일한 형태
데이터를 학습하려는데 편차가 너무크면 결과가 이상해져서 지금 배우는게 확률분포인듯


베르누이 분포를 n번실행하면 -> 이항분포 또 n번실행하면 -> 정규분포
-> 표준분포(평균 O, 표준편차 1인 정규분포)

정규분포 -- 표준화(평균0, 편차 1) --> 표준 정규분포

평균 : 데이터를 더한 후 데이터의 갯수로 나눈 값(데이터의 중심)

편차: 데이터의 값 - 평균 (모든 편차를 다 더하면 0)

분산: 편차를 제곱한 값들의 평균

표준편차: 평균에서 얼마나 떨어져 있느냐?

표준편차가 크면 =퍼져있다 작으면 = 모여있다.

```python
data = [2, 4, 8, 10]
평균 = 24 / 4
print(평균)
편차 = 2 - 6 , 4 - 6, 8- 6 , 10 -6
분산 =  40/4 #= 10-4^2 + 2^2 + 2^2 + 4^2 =
print(분산)
표준편차 = 3.16 # 10루트
```

```python
def array_info(array):
  print(array)
  print("ndim:", array.ndim)
  print("shape:", array.shape)
  print("dtype:", array.dtype)
  print("size:", array.size)
  print("itemsize:", array.itemsize)
  print("nbytes:", array.nbytes) #전체의 크기  itemsize 입력크기 * size 크기
  print("strides:", array.strides)
```

```python
array = np.array([2, 4, 6, 8])
array_info(array)
```

```python
#넘파이 산술연산
a = [21, 22, 23, 24, 25]
b = [11, 12, 13 ,14 ,15]

dap = np.add (a, b) #두배열의 값을 더한다
print(dap)

dap1 = np.subtract(a,b) # 두배열의 값을 뺀다
print(dap1)

dap2 = np.multiply(a,b) #두배열의 값을 곱한다.
print(dap2)

dap3 = np.divide(a,b) #값을 나눈다
print(dap3)

res = np.round(dap3, 2) #반올림
print(res)
```

```python
a = [10, 20, 30, 40, 50]
b = [1, 2, 3, 4, 5]
dap = np.mod(a,b) # 나머지
print(dap)

dap1 = np.divmod(a, b)
print(dap1) # 두개의 배열  첫번째는 몫, 두번째는 나머지를 출력
```

```python
a = [1,2,3,4,5]
b = [1,2,3,4,5]

dap = np.power(a,b)  # 첫번째 배열의 값을 두번째 배열 값으로 거듭제곱
print(dap)
```
