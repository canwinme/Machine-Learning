배열 변환
* 전치배열: 행과 열을 맞바꾸거나 다차원 배열의 축 순서를 변경한다.

```python
import numpy as np
```

```python
matrix = np.array([[1,2,3],[4,5,6]])
trans_matrix = matrix.T
print(trans_matrix)
```

* 브로드캐스팅 : 산술 연산이 되는 배열에서 모양(shape)이 다른 경우에도 연산이 가능하도록 배열의 모양을 변형
* 두차원이 같거나
* 둘 중 하나의 차원 크기가 1이거나
* 차원이 없으면 크기를 1로 취급

```python
a = np.array([1,2,3])
res = a + 2 #배열 + 스칼라 연산
print(res)
```

```python
A = np.array([[1,2,3,],[7,8,9]])
B = np.array([10,20,30])
res = A + B
print(A.shape)
print(B.shape)
print(res)
```

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[10], [20]])
print(a.shape)
print(b.shape)
print(a + b)
```

```python
A = np.ones([3, 4])
B = np.ones([1,3])
print(A.shape)
print(B.shape)
print(A+B)
```
