선형회귀와 의사결정나무 회귀 비교
- 선형회귀
  - 방식 : 데이터의 전반적인 경향을 나타내는 하나의 직선 y = wx +b를 찾습니다 (w: 가중치, b : y의 절편, x의 독립변수, y: 종속변수)
  - 특징 : 전체 영역에 대해 연속적이고 일정한 기울기를 가진다. 경향성이 강한 데이터에 유리하지만 복잡한 비선형 패턴을 표현하는데에는 한계가 있다.
- 의사결정 나무 회귀
  - 방식 : 특정 공간을 조건문으로 계속 분할하여 각 영역의 평균값으로 예측한다
  - 특징 : 국소적인 비선형 패턴을 잘잡아내며 계단 형태의 예측 곡선을 형성한다. 깊이가 깊어지면 데이터 하나하나에 적응하여 과적합을 발생하기 쉽다. (너무 많이 학습하는것도 좋지는 않다.)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# 비선형(2차원 곡선 + 노이즈) 데이터셋 만들기
np.random.seed(42)
X= np.sort(5 * np.random.rand(80, 1), axis = 0) #80에 1열짜리 2차원배열 생성 // 범위가 너무작기때문에 -> * 5
y = np.sin(X).ravel() + np.random.normal(0, 0.2, X.shape[0]) #X데이터에 사인함수를 적용, 평균이 0, 표준편차 0.2인 정규분포로 부터 무작위 오차를 80개 생성

#그래프 그리기
plt.figure(figsize = (10, 6))
plt.scatter(X, y, color = 'darkorange') #산점도 그리기
plt.plot(X, np.sin(X), color='navy', label='True(x) = sin(x)') #정답 곡선

plt.title("Dataset : y =sin(x)")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
lr_model = LinearRegression()
lr_model.fit(X,y)

#시험용 데이터 생성
X_test = np.linspace(0, 5, 500)[:, np.newaxis] #0 부터 5까지 500구간으로 쪼개고 차원을 늘려 2차원 배열로 변환
y_pred_lr = lr_model.predict(X_test)

#평가
y_train_pred_lr = lr_model.predict(X)
print(f"[선형회귀] Train MSE: {mean_squared_error(y, y_train_pred_lr):.4f}")
print(f"[선형회귀] Train R2: {r2_score(y, y_train_pred_lr):.4f}")
```

```python
#
from sklearn.tree import DecisionTreeRegressor
#제약이 없는 트리(과적합 위험)
df_full = DecisionTreeRegressor(random_state=42)
df_full.fit(X, y)
y_pred_dt_full = df_full.predict(X_test)

#깊이 추가
dt_reg3 = DecisionTreeRegressor(max_depth=3, random_state=42)
dt_reg3.fit(X, y)
y_pred_dt3 = dt_reg3.predict(X_test)

#평가
plt.figure(figsize=(12,6))
plt.scatter(X, y, color='black', label='Actual Data', alpha = 0.7)
plt.plot(X_test, y_pred_lr, color='blue',linewidth=2, label='Linear Regression')
plt.plot(X_test, y_pred_dt_full, color='red', linewidth='2', linestyle='--', label='Tree(max_depth=None)')
plt.plot(X_test, y_pred_dt3, color='green', linewidth=2, label='Decision Tree (Depth=3)')

plt.title("Comparison of Regression Models")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#모델 평가
models = {
    "Linear Regression": lr_model,
    "Decision Tree (max_depth=None)": df_full,
    "Decision Tree (max_depth=3)": dt_reg3
}
print("==========모델평가==========")
for name, model in models.items():
  preds = model.predict(X)
  mse = mean_squared_error(y, preds)
  r2 = r2_score(y, preds)
  print(f"[{name}] MSE: {mse:.4f} | R2 Score:{r2: .4f}")
```
