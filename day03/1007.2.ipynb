#인코딩


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler
```

```python
data = {
    'Age' : [25, 45, 35, 50, 23],
    'Income' : [30000, 85000, 60000, 120000, 280000],
    'Gender' : ['Male', 'Female', 'Female', 'Male', 'Female'],
    'Education' : ['Bachelor', 'Master', 'PhD', 'Bashelor', 'High School']
}

df = pd.DataFrame(data)
print(df)
```

1. 데이터 인코딩
* 라벨 인코딩: 연속적인 정수를 부여한다.
* 원핫 인코딩 : 0과 1의 이진벡터로 늘려 표현한다.

```python
#범주형 데이터 인코딩
#1. 라벨 인코딩

le = LabelEncoder()
df_encoded = df.copy()

df_encoded['Gender_Label'] = le.fit_transform(df['Gender'])

print("====== 라벨 인코딩 결과(Gender) ========")

print(df_encoded[['Gender', 'Gender_Label']])

#2. 원핫 인코딩
df_dummies = pd.get_dummies(df, columns= ['Education'], prefix = 'Edu', dtype= int)
print("===========원핫 인코딩==============")
print(df_dummies)
```

### 데이터 스케일링:
서로 다른 변수들의 데이터 범위를 일정한 기준으로 맞춰주는 전처리

* 표준화 : 데이터의 평균을 0, 분산을 1
* 정규화 : 데이터의 범위를 0과 1사이로 압축
* 로버스트 스케일링 : 평균 대신 중앙값과 사분위수를 사용해 스케일링

```python
from os import XATTR_SIZE_MAX
# 수치형 데이터 스케일링

df_num = df[['Age', 'Income']]
#1. 표준화 : 평균
scaler_std = StandardScaler()
X_std = scaler_std.fit_transform(df_num)
df_std = pd.DataFrame(X_std, columns = ['Age_Std', 'Income_Std'])
#2. 정규화
scaler_minmax = MinMaxScaler()
X_minmax = scaler_minmax.fit_transform(df_num)
df_minmax = pd.DataFrame(X_minmax, columns = ['Age_MinMax', 'Income_MinMax'])

df_scaled = pd.concat([df_num, df_std, df_minmax], axis = 1)
print("-----------------스케일링 적용비교-----------------")
print(df_scaled)
print("--------------스케일링 후 통계량 확인-----------------")
print("StandardScaler 평균 :", np.round(X_std.mean(axis=0),2))
print("StandardScaler 표준편차:",X_std.std(axis=0))
print("MinMaxScaler 최소값:", X_minmax.min(axis=0))
print("MinMaxScaler 최대값:", X_minmax.max(axis=0))
```

```python
#시각화
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
#원본 산점도
axes [0].scatter(df['Age'], df['Income'], color='blue', alpha = 0.7)
axes [0].set_title('Original Date')
axes [0].set_xlabel('Age')
axes [0].set_ylabel('Income')
#StandardScaler 산점도
axes [1].scatter(df_std['Age_Std'], df_std['Income_Std'], color='green', alpha = 0.7)
axes [1].set_title('StandardScaler (Mean=0, Std=1)')
axes [1].set_xlabel('Age(Std)')
axes [1].set_ylabel('Income(Std)')
#MinMaxScaler 산점도
axes [2].scatter(df_minmax['Age_MinMax'], df_minmax['Income_MinMax'], color='red', alpha = 0.7)
axes [2].set_title('MinMaxScaler (Range 0 ~ 1)')
axes [2].set_xlabel('Age(MinMax)')
axes [2].set_ylabel('Income(MinMax)')

plt.tight_layout()
plt.show()
```

```python
titanic = sns.load_dataset('titanic')

print("=========타이타닉데이터========")
print(titanic.head(10))

print ("======결측치 현황 탐색=======")
missing_counts = titanic.isnull().sum()
print(missing_counts)
print("=====컬럼별 결측치 개수=====")
print(missing_counts[missing_counts > 0])
print("=====컬럼별 결측치 비율(%)======")
missing_ratio = (missing_counts / len(titanic)) * 100
print(missing_ratio[missing_ratio > 0].round(2))
```

```python
# 결측치 삭제
# 결측치가 포함된 행 모두 삭제 : dropna()
df_droped_rows = titanic.dropna(subset = ['embarked']) #embarked가 NaN인 행 삭제
print(f"embarked행 삭제전 : {len(titanic)}행 --> 삭제후 : {len(df_droped_rows)} 행")

#결측치 비율이 너무 높은 열 삭제
df_cleaned = titanic.drop(columns = ['deck'])
print(f"deck 컬럼 삭제후 남은 컬럼 :", df_cleaned.columns.tolist())
```

```python
# 결측치 대체
age_median = df_cleaned['age'].median()
print(f"age 컬럼 중앙값: {age_median: .1f} 세")

df_cleaned['age'] = df_cleaned['age'].fillna(age_median)

embarked_mode = df_cleaned['embarked'].mode()[0]
print(f"embarked 컬럼 최빈값: {embarked_mode}")

df_cleaned['embarked'] = df_cleaned['embark_town'].fillna(embarked_mode)
df_cleaned['embark_town'] = df_cleaned['embark_town'].fillna(embarked_mode)

print("[정제 완료 후 결측치 확인]")
print(df_cleaned[['age','embarked','embark_town']].isnull().sum())
```
