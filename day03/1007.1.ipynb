```python
# 결측치 및 이상치가 포함된 데이터 셋
np.random.seed(42)
n_samples = 100

df = pd.DataFrame({
    '평수': np.random.randint(10, 60, n_samples).astype(float),
    '방개수': np.random.randint(1, 5, n_samples).astype(float),
    '건축연도': np.random.randint(2000 ,2024, n_samples).astype(float),
    '집값': np.random.randint(10000, 50000, n_samples).astype(float)
})
df.head()
#의도적 결측치
df.loc[5:8, '방개수'] = np.nan
df.loc[20:22, '평수'] = np.nan
#의도적 이상치
df.loc[50, "평수"] =990.0
df.loc[70, "집값"] = -9999.0

print("========결측치/이상치 데이터 확인========")
print(df.head(10))
```

```python
#결측치 확인 및 처리
#결측치 삭제 혹은 대체로 해결가능
print("==== 컬럼별 결측치 확인=====  ")
print(df.isnull().sum())

#1. 삭제(dropna)
df_drop = df.dropna()
print(f"\n 원본 크기: {df.shape} -> 삭제 후 크기: {df_drop.shape}")
print()
#2. 대체(fillna)
df_clean = df.copy()
df_clean['평수'] = df_clean['평수'].fillna(df_clean['평수'].median())
df_clean['방개수'] = df_clean['방개수'].fillna(df_clean['방개수'].median())
print("===결측치 대체후 null 확인===")
print(df_clean.isnull().sum())
```

*  2. 이상치 처리
* 이상치 처리 이상치 IQR 공식:



```python

Q1 = df_clean['평수'].quantile(0.25) # 하위 25%지점 찾기
Q3 = df_clean['평수'].quantile(0.75) # 하위 75%지점 찾기
IQR = Q3 - Q1
#범위 밖에 있는값이면 이상치
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"평수 Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"이상치 하한선: {lower_bound: 2f}, 상한선: {upper_bound:.2f}")

outliers = df_clean[(df_clean['평수']<lower_bound)| (df_clean['평수']>upper_bound)]
print("=========발견된 이상치 데이터 =========")
print(outliers[['평수','방개수','집값']])
```

* 3.이상치 제거 또는 캡핑

```python
df_clean['평수_정제'] = df_clean['평수'].clip(lower=lower_bound, upper = upper_bound)

df_final = df_clean[df_clean['집값']> 0].copy()

print("=========최종 정제 후 기술 통계량========")
df_final[['평수_정제', '방개수','집값']].describe().round(2)
```

* 결측치는 함부로 삭제하면 데이터 손실이 크므로 대체를 우선 검토해야한다.

* 이상치는 무조건 지우기보다 극단값인지 구분을 해야 한다.

* 이상치가 포함되어 있는 경우는 평균이 왜곡되기 쉬우므로 중값으로 대체하는 것이 좋다.
